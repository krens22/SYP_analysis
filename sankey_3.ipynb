# FILTER 2 - last occurance

```python
# Sort by ID and YEAR to get the latest entry for each student
filter2 = csv_norm_2015_2023_ug.sort_values(['ID', 'YEAR'], ascending=[True, True])
```

```python
# Group by ID and get the last entry for each student (this will be their final major)
final_major_df = filter2.groupby('ID').last().reset_index()
final_major_df
```

```python
final_major_df['PROGRAM'].unique()
```

```python
programs_to_keep = ['SYP', 'ESP', 'WIE', 'NST', 'YMT', 'WCS', 'YPP', 'GU', 'YGW', 'YPO']

def filter_by_program(program):
    if program not in programs_to_keep:
        return f"Program {program} is not in the allowed list."
    return final_major_df[final_major_df['PROGRAM'] == program]
```

```python
import pandas as pd
import plotly.graph_objects as go

# Assuming the data is already loaded in csv_norm_2015_2023_ug
# Node 1: Group by 'LEVEL' and 'DEPARTMENT'
san_year_level = filter_by_program('SYP').groupby(['GENDER', 'PROGRAM'])['ID'].count().reset_index()
san_year_level = san_year_level.rename(columns={'GENDER': 'source', 'PROGRAM': 'target', 'ID': 'value'})

# Node 2: Group by 'DEPARTMENT' and 'PROGRAM'
san_year_program = filter_by_program('SYP').groupby(['PROGRAM', 'DEPARTMENT'])['ID'].count().reset_index()
san_year_program = san_year_program.rename(columns={'PROGRAM': 'source', 'DEPARTMENT': 'target', 'ID': 'value'})

# Combine the two datasets
combined_df = pd.concat([san_year_level, san_year_program])

# Unique values that contain unique values of source and target
unique_nodes = list(pd.unique(combined_df[['source','target']].values.ravel("K")))

# Dictionary to map nodes to indices
mapping_dict = {k: v for v, k in enumerate(unique_nodes)}

# Map the nodes in the DataFrame
combined_df['source'] = combined_df['source'].map(mapping_dict)
combined_df['target'] = combined_df['target'].map(mapping_dict)

# Convert DataFrame to a dictionary with lists
link_dict = combined_df.to_dict(orient="list")

# Step 1: Calculate counts for each node
node_counts = combined_df.groupby('source')['value'].sum().to_dict()
node_counts.update(combined_df.groupby('target')['value'].sum().to_dict())

# Step 2: Create a new label with counts
updated_labels = [f"{node} ({node_counts.get(idx, 0)})" for idx, node in enumerate(unique_nodes)]

# Step 3: Create the Sankey diagram with updated labels
fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=updated_labels,  # Use updated labels here
    ),
    link=dict(
        source=link_dict["source"],
        target=link_dict["target"],
        value=link_dict["value"],
    )
)])

# Update layout
fig.update_layout(title_text="Sankey Diagram by first department", font_size=10)

# Display the Sankey diagram
fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

# Assuming the data is already loaded in csv_norm_2015_2023_ug
# Node 1: Group by 'LEVEL' and 'DEPARTMENT'
san_year_level = filter_by_additional_program('Mechanical Engineering').groupby(['DEPARTMENT', 'PROGRAM'])['ID'].count().reset_index()
san_year_level = san_year_level.rename(columns={'DEPARTMENT': 'source', 'PROGRAM': 'target', 'ID': 'value'})

# Node 2: Group by 'DEPARTMENT' and 'PROGRAM'
san_year_program = filter_by_additional_program('Mechanical Engineering').groupby(['PROGRAM', 'GENDER'])['ID'].count().reset_index()
san_year_program = san_year_program.rename(columns={'PROGRAM': 'source', 'GENDER': 'target', 'ID': 'value'})

# Combine the two datasets
combined_df = pd.concat([san_year_level, san_year_program])

# Unique values that contain unique values of source and target
unique_nodes = list(pd.unique(combined_df[['source','target']].values.ravel("K")))

# Dictionary to map nodes to indices
mapping_dict = {k: v for v, k in enumerate(unique_nodes)}

# Map the nodes in the DataFrame
combined_df['source'] = combined_df['source'].map(mapping_dict)
combined_df['target'] = combined_df['target'].map(mapping_dict)

# Convert DataFrame to a dictionary with lists
link_dict = combined_df.to_dict(orient="list")

# Step 1: Calculate counts for each node
node_counts = combined_df.groupby('source')['value'].sum().to_dict()
node_counts.update(combined_df.groupby('target')['value'].sum().to_dict())

# Step 2: Create a new label with counts
updated_labels = [f"{node} ({node_counts.get(idx, 0)})" for idx, node in enumerate(unique_nodes)]

# Step 3: Create the Sankey diagram with updated labels
fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=updated_labels,  # Use updated labels here
    ),
    link=dict(
        source=link_dict["source"],
        target=link_dict["target"],
        value=link_dict["value"],
    )
)])

# Update layout
fig.update_layout(title_text="Sankey Diagram by first department", font_size=10)

# Display the Sankey diagram
fig.show()
```
